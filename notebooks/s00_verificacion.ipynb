```python
# Verificación de entorno · ejecutar primero
import sys
import platform

print("Python:", sys.version.split()[0])
print("Sistema:", platform.system(), platform.release())

required_packages = ["pandas", "numpy", "psutil"]
missing_packages = []

for package_name in required_packages:
    try:
        __import__(package_name)
        print(f"OK ·{package_name}")
    except ImportError:
        missing_packages.append(package_name)
        print(f"FALTA ·{package_name}")

if missing_packages:
    print("\nInstale con:  pip install " + " ".join(missing_packages))
else:
    print("\nEntorno listo.")
```

```python
# Cuaderno de verificacion del entorno
import sys, pandas, numpy, psutil

print("Python:", sys.version.split()[0])
print("pandas:", pandas.__version__)
print("numpy:", numpy.__version__)

# Verificar que las versiones coinciden con lo anclado
expected = {"pandas": "2.2.2", "numpy": "1.26.4"}
actual = {"pandas": pandas.__version__, "numpy": numpy.__version__}

for package, version in expected.items():
    status = "OK" if actual[package] == version else "DISCREPANCIA"
    print(f"{status} ·{package}: esperado{version}, obtenido{actual[package]}")

print("\nEntorno reproducible verificado." if actual == expected
      else "\nRevise el anclaje de versiones.")
```
